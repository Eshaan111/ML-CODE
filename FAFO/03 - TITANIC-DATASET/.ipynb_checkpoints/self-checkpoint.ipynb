```python
from sklearn.model_selection import StratifiedShuffleSplit

splitter = StratifiedShuffleSplit(
    n_splits= 1,
    test_size=0.2,
    random_state=23
)

for train_indices,test_indices in splitter.split(df, df[['Survived',"Sex","Pclass"]]) :

    strat_train = df.loc[train_indices]
    strat_test = df.loc[test_indices]

strat_train.shape, strat_test.shape
```

# IMPUTER

```python
from sklearn.impute import SimpleImputer
from sklearn.base import BaseEstimator, TransformerMixin

class AgeImputer(BaseEstimator, TransformerMixin):
    def __init__(self):
        self.imputer = SimpleImputer(strategy="mean")

    def fit(self, X, y=None):
        self.imputer.fit(X[["Age"]])
        return self

    def transform(self, X):
        X = X.copy()
        X["Age"] = self.imputer.transform(X[["Age"]]).ravel()
        return X


```

# ENCODER

```python
from sklearn.preprocessing import OneHotEncoder
from sklearn.compose import ColumnTransformer

class ParaEncoder(BaseEstimator, TransformerMixin):
    def __init__(self):
        self.cat_features = ["Embarked", "Sex"]

        self.transformer = ColumnTransformer(
            transformers=[
                (
                    "one_hot",
                    OneHotEncoder(
                        handle_unknown="ignore",
                        sparse_output=False
                    ),
                    self.cat_features
                )
            ],
            remainder="passthrough",
            verbose_feature_names_out=False
        )

        self.transformer.set_output(transform="pandas")

    def fit(self, X, y=None):
        self.transformer.fit(X)
        return self

    def transform(self, X):
        return self.transformer.transform(X)

tempX = strat_train
encode_obj = ParaEncoder()
tempX = encode_obj.fit_transform(X = tempX)
pd.DataFrame(tempX)
```

# Feature Dropper


```python
class Dropper(BaseEstimator, TransformerMixin): 
    def __init__(self) :
        return None
    def fit(self, X, y=None):
        self.is_fitted_ = True
        return self
    
    def transform(self,X) :

        return X.drop(["Ticket","Cabin","Name","Embarked_nan"], axis = 1, errors = "ignore")
```

# PIPELING

```python
from sklearn.pipeline import Pipeline

pipeline = Pipeline(
    [
        ("AgeImputer",AgeImputer()),
        ("Encoder", ParaEncoder()),
        ("FeatureDropper",Dropper())
    ]
)


```

```python
strat_train.info()
```

```python
strat_train_set = pipeline.fit_transform(strat_train)
```

```python
strat_train_set.info()
```

# STANDARDISING

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()



X = strat_train_set.drop(["Survived"], axis=1)
y = strat_train_set["Survived"]

X_data = scaler.fit_transform(X)
y_data = y.to_numpy()



strat_test_set = pipeline.transform(strat_test)

X_test_data = strat_test_set.drop(['Survived'], axis =1)
y_test_data = strat_test_set['Survived']

X_test_data = scaler.transform(X_test_data)
y_test_data = y_test_data.to_numpy()
```

# MODEL FITTING


```python
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    confusion_matrix
)
```

## RANDOM FOREST


```python
from sklearn.ensemble import RandomForestClassifier
from sklearn.model_selection import GridSearchCV

clf = RandomForestClassifier()

param_grid = [{
    "n_estimators" : [10,100,200,500],
    "max_depth" : [None,5,10],
    "min_samples_split" : [2,3,4]
}]

grid_search = GridSearchCV(clf, param_grid=param_grid, scoring="accuracy", return_train_score=True, n_jobs=-1)
grid_search.fit(X_data,y_data)
```

USING THE SMAE SCALER WE FIT ON TRAINIGN DATA

```python


strat_test_set = pipeline.transform(strat_test)

X_test = strat_test_set.drop(["Survived"], axis=1)
y_test = strat_test_set["Survived"]

X_test_data = scaler.transform(X_test)
y_test_data = y_test.to_numpy()

X_test_data

```

```python
best_clf = grid_search.best_estimator_
best_clf.score(X_test_data,y_test_data)
y_pred_rf = best_clf.predict(X_test_data)
```

```python
print("Confusion matrix:")
print(confusion_matrix(y_test_data, y_pred_rf))

print("Accuracy: ", accuracy_score(y_test_data, y_pred_rf))
print("Precision:", precision_score(y_test_data, y_pred_rf))
print("Recall:   ", recall_score(y_test_data, y_pred_rf))
print("F1:       ", f1_score(y_test_data, y_pred_rf))
```
